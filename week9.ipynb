```python
import pandas as pd

# =========================
# 📊 LOAD DATA
# =========================

sales_df = pd.read_csv("sales_data.csv")
transactions_df = pd.read_csv("transactions.csv")

print("\n📊 DATA LOADED SUCCESSFULLY\n")

# =========================
# 🧹 FORCE NUMERIC CLEANING
# =========================

sales_df['Total_Sale'] = pd.to_numeric(
    sales_df['Total_Sale'].astype(str).str.replace(r'[\$,]', '', regex=True),
    errors='coerce'
)

sales_df['Unit_Price'] = pd.to_numeric(
    sales_df['Unit_Price'].astype(str).str.replace(r'[\$,]', '', regex=True),
    errors='coerce'
)

transactions_df['Amount'] = pd.to_numeric(
    transactions_df['Amount'].astype(str).str.replace(r'[\$,]', '', regex=True),
    errors='coerce'
)

# =========================
# 📊 SALES DATA ANALYSIS
# =========================

# Clean currency columns properly
sales_df['Unit_Price'] = sales_df['Unit_Price'].replace(r'[\$,]', '', regex=True).astype(float)
sales_df['Total_Sale'] = sales_df['Total_Sale'].replace(r'[\$,]', '', regex=True).astype(float)

print("\n================ SALES OVERVIEW ================\n")

# 1. Sales by Region
region_summary = sales_df.groupby('Region').agg(
    total_sales=('Total_Sale', 'sum'),
    avg_sale=('Total_Sale', 'mean'),
    transactions=('Total_Sale', 'count')
).reset_index()

region_summary = region_summary.sort_values(by='total_sales', ascending=False)

print("📍 SALES BY REGION:")
print(region_summary)


# 2. Product Analysis
product_summary = sales_df.groupby('Product').agg(
    total_revenue=('Total_Sale', 'sum'),
    total_units=('Quantity', 'sum')
).reset_index()

print("\n🛍️ TOP 5 PRODUCTS BY REVENUE:")
print(product_summary.sort_values(by='total_revenue', ascending=False).head(5))

print("\n📦 TOP 5 PRODUCTS BY UNITS SOLD:")
print(product_summary.sort_values(by='total_units', ascending=False).head(5))


# 3. Sales Rep Performance
rep_summary = sales_df.groupby('Sales_Rep').agg(
    total_revenue=('Total_Sale', 'sum'),
    avg_sale=('Total_Sale', 'mean'),
    transactions=('Total_Sale', 'count')
).reset_index()

rep_summary = rep_summary.sort_values(by='total_revenue', ascending=False)

print("\n👨‍💼 SALES REP PERFORMANCE:")
print(rep_summary)


# 4. Region + Sales Rep Analysis
region_rep = sales_df.groupby(['Region', 'Sales_Rep']).agg(
    total_sales=('Total_Sale', 'sum')
).reset_index()

print("\n🌍 REGION + SALES REP PERFORMANCE:")
print(region_rep)


# =========================
# 💳 TRANSACTION ANALYSIS
# =========================

print("\n================ TRANSACTIONS OVERVIEW ================\n")

# Convert Date column to datetime
transactions_df['Date'] = pd.to_datetime(transactions_df['Date'])

# Add Month and Day columns
transactions_df['Month'] = transactions_df['Date'].dt.to_period('M')
transactions_df['DayOfWeek'] = transactions_df['Date'].dt.day_name()


# 1. Monthly Spending Trend
monthly_spending = transactions_df.groupby('Month')['Amount'].sum()

print("📈 MONTHLY SPENDING TREND:")
print(monthly_spending)


# 2. Spending by Day of Week
weekday_spending = transactions_df.groupby('DayOfWeek')['Amount'].mean()

print("\n📅 AVERAGE SPENDING BY DAY OF WEEK:")
print(weekday_spending)


# 3. Category Spending Analysis
category_spending = transactions_df.groupby('Category')['Amount'].agg(
    ['sum', 'mean', 'count']
)

print("\n💰 CATEGORY SPENDING ANALYSIS:")
print(category_spending)


# 4. Spending Variation
category_variation = transactions_df.groupby('Category')['Amount'].std()

category_variation = category_variation.sort_values(ascending=False)

print("\n⚠️ SPENDING VARIATION BY CATEGORY:")
print(category_variation)


# 5. Suggested Monthly Budget
monthly_category = transactions_df.groupby(
    ['Month', 'Category']
)['Amount'].sum().reset_index()

avg_budget = monthly_category.groupby('Category')['Amount'].mean()

print("\n📊 SUGGESTED MONTHLY BUDGET BY CATEGORY:")
print(avg_budget)


# =========================
# ✅ COMPLETED
# =========================

print("\n✅ ALL ANALYSES COMPLETED SUCCESSFULLY")

#MY PROMPTS
# 1. Analyze sales by region and show total sales, average sales, and transaction counts.
# 2.Find the top 5 products by revenue and units sold.
# 3.Analyze sales rep performance using revenue and average sales.
# 4. Analyze spending trends by date and day of week.
# 5. Calculate average spending by category and suggest budgets/
# 6. Simplify this code and make it easier to understand.

# Reflection : I learned how to install python library like pandas, numpy. I also learned how to clean and organize CSV data using Pandas, analyze sales and transaction datasets, and generate useful business insights such as top-performing regions, products, and spending patterns.
```
