```python
try: #tries to open the file
    with open("my.data.txt", "r") as file:
        #initialize all to 0
        char_count = 0
        word_count = 0
        line_count = 0
        longest_line = ""

        #looping through all the lines
        for line in file:
            line_count += 1
            char_count += len(line)
            words = line.split()
            word_count += len(words)

            #to find the longest line present in the file
            if len(line) > len(longest_line):
                longest_line = line #stores the longest line

        print("File Analysis Report: ")
        print("Total characters: ", char_count)
        print("Total words: ", word_count)
        print("Total lines: ", line_count)
        print("Longest line: ", longest_line)

#if file doesn't exist
except FileNotFoundError:
    print("File is not found!")



```
