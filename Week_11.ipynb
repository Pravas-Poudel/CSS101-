```python
#MODULE 1
Celsius to Fahrenheit Converter:
1. Plain Language Description
We need to create a function called celsius_to_fahrenheit() that takes a temperature in Celsius and converts it into Fahrenheit using the formula:
                    F=9/5*C + 32
The function should return the converted temperature.

2. Example Inputs and Output
The input will be celsius temperature. Example : 0, 32
The output will be fahrenheit temperature. Example : 32, 212

3. Edge Cases
a. Very low temperature
    Input: -273.15
    Output: -459.67
    Reason: Absolute zero

b. Decimal values
    Input: 36.6
    Output: 97.88
    Reason: Function should handle floating point numbers

AI Implementation (Python):
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

Verify with Assertions:
assert celsius_to_fahrenheit(0) == 32
assert celsius_to_fahrenheit(100) == 212
assert celsius_to_fahrenheit(25) == 77
assert celsius_to_fahrenheit(-40) == -40
assert round(celsius_to_fahrenheit(-273.15), 2) == -459.67
print("All tests passed!")

Simplified Version:
def c_to_f(c):
    return c * 1.8 + 32
```

```python
#MODULE 2
1. Expected vs Actual Behavior
Expected Behavior
The function should count all vowels letter no matter if it is uppercase or lowercase vowels.
Example:
count_vowels("Hello")
Expected output:
2
Another example:
count_vowels("HELLO")
Expected output:
2

Actual Behavior
The function only counts lowercase vowels because:
vowels = "aeiou"
So uppercase vowels like "E" or "O" are ignored.
Example:
count_vowels("HELLO")
Actual output:
0

2. Add Prints to Trace Execution
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for letter in text:
        print("Checking:", letter)

        if letter in vowels:
            print(letter, "is a vowel")
            count += 1

    print("Final count:", count)
    return count

Example run:
count_vowels("Hello")
Trace:
Checking: H
Checking: e
e is a vowel
Checking: l
Checking: l
Checking: o
o is a vowel
Final count: 2

3. Form Hypothesis About the Cause
Hypothesis:
The function only checks lowercase vowels.
Uppercase vowels are not being matched.

4. Identify the Root Cause
Root cause:
if letter in vowels:
The variable vowels only contains:
"aeiou"
So:
"E" in "aeiou"
returns:
False

5. Implement Simple Fix
Convert each letter to lowercase before checking.
def count_vowels(text):
    vowels = "aeiou"
    count = 0

    for letter in text:
        if letter.lower() in vowels:
            count += 1

    return count

6. Verify with Multiple Test Cases
assert count_vowels("Hello") == 2
assert count_vowels("HELLO") == 2
assert count_vowels("AEIOU") == 5
assert count_vowels("Python") == 1
assert count_vowels("") == 0
assert count_vowels("xyz") == 0
assert count_vowels("Beautiful") == 5

print("All tests passed!")
```

```python
2. Input and Output Examples:
Valid cases:
1234567890 → (123) 456-7890
123-456-7890 → (123) 456-7890
123 456 7890 → (123) 456-7890
(123)4567890 → (123) 456-7890
(123) 456 7890 → (123) 456-7890

Invalid cases:
123456789 → Error: Invalid phone number
12345678901 → Error: Invalid phone number
12a4567890 → Error: Invalid characters
`` → Error: Input required
null → Error: Input required
```

```python
3. Edge Cases / Boundaries:
Exactly 10 digits will be valid
9 digits or less will be invalid
11+ digits will also be invalid

Strange formats:
"123 456 7890" (extra spaces)
"123-456--7890" (double separators)
"(123) 456-7890" (extra spaces)
"123.456.7890" (dots instead of dashes)
```

```python
4. Invalid Input Handling
The function must reject:
Inputs with wrong digit count (not 10 digits)
Expected Error : Invalid phone number.

Inputs with letters (abc1234567)
Expected Error : Invalid characters.

Inputs with symbols (@#$%^)
Expected Error : Invalid characters.

Empty input ("")
Expected Error : Input required.

Null input
Expected Error : Input required.
```

```python
def format_phone(phone):
    if phone is None or phone == "":
        return "Error"

    # extract digits only
    digits = "".join(ch for ch in str(phone) if ch.isdigit())

    # must be exactly 10 digits
    if len(digits) != 10:
        return "Error"

    # format
    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"

# TESTS
assert format_phone("1234567890") == "(123) 456-7890"
assert format_phone("123-456-7890") == "(123) 456-7890"
assert format_phone("123 456 7890") == "(123) 456-7890"
assert format_phone("(123)4567890") == "(123) 456-7890"

assert format_phone("123456789") == "Error"
assert format_phone("12a4567890") == "Error"
assert format_phone("") == "Error"
assert format_phone(None) == "Error"

print("All tests passed!")
```
