```python
#importing libraries
import pandas as pd
import pyinputplus as pyip

#Creating a menu
def show_menu():
    print("\n==== Personal Budget Tracker ====")
    print("1. Add Transaction")
    print("2. View All Transactions")
    print("3. View Current Balance")
    print("4. Save and Exit")

    choice = pyip.inputInt(
    prompt="Enter your choice:",
    min=1,
    max=4
    )
    return choice

#Create add transaction function
def add_transaction(df):
    date = pyip.inputStr(
        prompt="Enter date (YYYY-MM-DD):"
    )
    description=pyip.inputStr(
        prompt="Enter description:"
    )
    amount = pyip.inputFloat(
        prompt="Enter amount: $",
        min = 0.01
    )
    transaction_type=pyip.inputChoice(['income', 'expense'],
                                      prompt="Enter transaction type:"
                                      )

    new_transaction ={'date': date,
                      'description':description,
                      'type': transaction_type,
                      'amount': amount
    }
    df.loc[len(df)] = new_transaction
    df.to_csv("budget.csv", index=False)
    print("Transaction saved successfully!")
    return df

#Create function to calculate the total balance remained in the account
def calculate_balance(df):
    income = df[df['type'] == 'income']['amount'].sum()
    expense = df[df['type'] == 'expense']['amount'].sum()
    return income - expense

test_df = pd.DataFrame({ 'date':['2025-01-01','2025-01-02'],
                         'description':['Salary','Rent'],
                         'type':['income','expense'],
                         'amount':[1000,200]
                         })
assert calculate_balance(test_df) == 800

#File Handling
try:
    budget_df = pd.read_csv("budget.csv")
    print("Budget data loaded successfully!")
except FileNotFoundError:
    budget_df = pd.DataFrame(
        columns=['date','description','type','amount']
    )
    print("No existing budget file found!")


while True:
    choice = show_menu()
    if choice == 1:
        budget_df = add_transaction(budget_df)
    elif choice == 2:
        if budget_df.empty:
            print("No transactions found!")
        else:
            print(budget_df)

    elif choice == 3:
        balance = calculate_balance(budget_df)
        print(f"Current Balance: ${balance:.2f}")

    elif choice == 4:
        budget_df.to_csv("budget.csv", index=False)
        print("Goodbye!")
        break




```
