```python
#WEEK 8 FINANCE MANAGER PROJECT

def load_transactions(filename):
    #Read lines from "transaction.txt" file safely.
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()
        return[line.strip() for line in lines if line.strip()]
    except FileNotFoundError:
        print(f"File '{filename}' not found!")
        return []

def process_transactions(lines):
#Process transaction lines into clean data.
    transactions = []
    failed_lines = 0

    for line in lines:
        try:
            # Split by comma
            parts = line.split(",")

            # Check if line has exactly 2 parts
            if len(parts) != 2:
                raise ValueError("Invalid format")

            description = parts[0]
            amount = float(parts[1])  # convert to number

            # Create dictionary
            transaction = {
                "description": description,
                "amount": amount
            }

            transactions.append(transaction)

        except:
            # If anything goes wrong, skip line
            print(f"Skipping bad line: {line}")
            failed_lines += 1

    print(f"\n✔ Successfully processed: {len(transactions)} transactions")
    print(f"✖ Failed lines: {failed_lines}\n")

    return transactions


def show_summary(transactions):
    #Display results to the user.
    total_income = 0
    total_expense = 0

    for t in transactions:
        amount = t["amount"]

        if amount > 0:
            total_income += amount
        else:
            total_expense += amount

    net = total_income + total_expense

    print("📊 Summary")
    print("-" * 30)
    print(f"Total transactions: {len(transactions)}")
    print(f"Total income: ${total_income:.2f}")
    print(f"Total expenses: ${total_expense:.2f}")
    print(f"Net amount: ${net:.2f}")


# Main program
def main():
    print("🏦 Personal Finance Manager")
    print("=" * 30)

    filename = "transactions.txt"
    raw_lines = load_transactions(filename)

    if not raw_lines:
        print("No data to process!")
        return

    transactions = process_transactions(raw_lines)
    show_summary(transactions)

main() #Without this the program doesnt run.


""" What was the hardest part and how did AI help me?
-> The hardest part while doing this project was dealing with bad data without crashing my program. When there was a wrong line my program would crash and stopped working. I then asked AI for help and it taught me how to use try except so that the program skips the bad line and doesnt crash. AI also helped me to understand how to convert text into integer and show the data properly. Because of this my program can handle error easily and runs better."""
```
