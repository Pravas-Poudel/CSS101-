```python

******************************************************************
PROBLEM 1 : Tip Calculator
******************************************************************
STEP 1 : Understand the problem
This program is used to help the staff of restaurant to calculate tips quickly and accurately. The staff enters the bill amount and tip percentage. After that the program calculates both the tip amount and the final total amount the customer needs to pay.
******************************************************************
******************************************************************
STEP 2: Identify inputs & outputs
INPUTS:
1. Bill amount entered by user.
Data type: Float
Example: $20.05
2. Tip percentage entered by user.
Data type: Integer or Float number
Example: 15% or 15.15%
******************************************************************
OUTPUTS:
1. Tip amount
Data type: Float
Example: $7.21
2. Final total amount
Data type: Float
Example: $21.21
******************************************************************
******************************************************************
STEP 3: Work it by hand
Example:
- Bill amount = $45.80
- Tip percentage = 15%
******************************************************************
Steps:
1. Convert 15% into decimal form:
    15% = 0.15
2. Multiply bill amount by tip percentage:
    $45.80 × 0.15 = $6.87
3. Add the tip amount to the original bill:
    $45.80 + $6.87 = $52.67
4. Final answer to the staff member:
“The tip amount is $6.87 and the total charge is $52.67.”
******************************************************************
******************************************************************
******************************************************************

******************************************************************
PROBLEM 2: Late Fee Calculator
******************************************************************
STEP 1: Understand the Problem
This program helps library staff to calculate overdue fines for the books burrowed. The librarian enters the number of days the book is overdue and the program calculates the fine based on $0.50 per day. It has a maximum fine limit of $10.00.
******************************************************************
STEP 2: Identify Inputs and Outputs
INPUTS:
1. Number of overdue days
Data type: Integer
Example: 1,2,3
******************************************************************
OUTPUTS:
1. Fine amount
Data type: Float
Example: $2.50
******************************************************************
2. Message: “Maximum fine reached”
Data type: String
******************************************************************
STEP 3: Work it by hand.
******************************************************************
Case A: When book is 5 days late,
My calculation:
5 × $0.50 = $2.50
For this case, the fine would be $2.50.
******************************************************************
Case B: When book is 25 days late,
My calculation:
25 × $0.50 = $12.50
The calculated fine is more than the maximum allowed fine of $10.00.
So, for this case the fine would be $10.00 as it is the maximum allowed fine.

Special note:
The fine is capped at the maximum limit so the librarian would see “Maximum fine reached.” in the display.
******************************************************************
Case C: When book is 0 days late,
For this case, there wont be any fine because the book is returned in time.
******************************************************************


                          THANKYOU

```
